Chi-Square calculation

```python
def DELTA_func(ylist,PRED):
    l = len(ylist)
    theory = np.zeros(l)
    for i in range(l):
        simpson = 1/12*(PRED[0][i] + 4*PRED[1][i] + 2*PRED[2][i] + 4*PRED[3][i] + PRED[4][i])
        theory[i] = np.nan_to_num(simpson,nan=0.0)
    return theory - ylist

def CHI2_func(DELTA,ERROR):
    chi_squared = np.sum(np.square(DELTA/ERROR))
    return chi_squared
```

Define Objective

```python
def objective(params, df, ratio_list):

    N = 5
    
    chi2 = 0
    N_total = 0
    
    αs = params[0]
    params_noαs = params[1:]

    μJ_ratio=float(ratio_list["μJ_ratio"])
    μH_ratio=float(ratio_list["μH_ratio"])
    μren_ratio=float(ratio_list["μren_ratio"])    

    Main.eval(f'@everywhere αs_Z = {αs}')
    Main.eval(f'@everywhere params = {params_noαs}') 
    Main.eval(f'@everywhere μJ_ratio = {μJ_ratio}')
    Main.eval(f'@everywhere μH_ratio = {μH_ratio}')
    Main.eval(f'@everywhere μren_ratio = {μren_ratio}') 

    dataset_names = df.keys()

    for name in dataset_names:
        xlist={}
        xlist_mid = np.array(df[name]["CHI"])
        binsize = np.array(df[name]["BinSize"]) 
        for i in range(N):
            xlist[i]=xlist_mid-binsize/2+i*binsize/(N-1)

        ylist=np.array(df[name]["EEC"])
        ERROR_list=np.array(df[name]["ERROR"])

        data_size = len(xlist_mid)
        N_total = N_total + data_size

        PRED={}

        parameters = params_noαs
        for quartile in range(N):
            PRED[quartile] = Main.model(xlist=xlist[quartile], Qlist=np.array(df[name]["Q"]))   
                         
        DELTA =  DELTA_func(ylist,PRED)                       

        chi_square = CHI2_func(DELTA,ERROR_list)
        chi2 = chi2 + chi_square

    return chi2/N_total
```

```python
import sys

class ProgressCallback:
    def __init__(self, df, ratio_list):
        self.df = df
        self.dataset_names = dataset_names
        self.ratio_list = ratio_list

    def __call__(self, params):
        chi2_bydof = objective(params, self.df, self.ratio_list)
        sys.stdout.flush()

class MyBounds:
    def __init__(self, xmax, xmin):
        self.xmax = np.array(xmax)
        self.xmin = np.array(xmin)

    def __call__(self, **kwargs):
        x = kwargs["x_new"]
        tmax = bool(np.all(x <= self.xmax))
        tmin = bool(np.all(x >= self.xmin))
        return tmax and tmin
```

```python
import os
import json
import numpy as np
import pandas as pd
from tqdm import tqdm

# Hard global bounds for [αs, a1, a2, a3, a4]
global_bounds = np.array([
    [0.118, 0.118],
    [0.7, 1.0],
    [1.5, 3.0],
    [0.9, 1.1],
    [1.1, 1.4],
], dtype=float)

center_params = np.array(initial_params, dtype=float)
sample_width = np.array([0.0, 0.004, 0.28, 0.029, 0.023], dtype=float)

out_path = "result/variation.csv"
cols = ["αs", "params", "chi2/dof", "muH_ratio", "muJ_ratio", "muRen_ratio"]

def sample_gaussian_bounded(center, width, bounds):
    x = center.copy()
    for j, (lo, hi) in enumerate(bounds):
        if width[j] <= 0 or lo == hi:
            x[j] = center[j]
            continue
        while True:
            v = np.random.normal(center[j], width[j])
            if lo <= v <= hi:
                x[j] = v
                break
    return x

# Create file with header once
if not os.path.exists(out_path) or os.path.getsize(out_path) == 0:
    pd.DataFrame(columns=cols).to_csv(out_path, index=False)

# Fixed inputs
df_fixed = df_list[0]
ratio_list = {"μH_ratio": 1.0, "μJ_ratio": 1.0, "μren_ratio": 1.0}

n_to_add = 9000
for _ in tqdm(range(n_to_add)):
    sampled_params = sample_gaussian_bounded(center_params, sample_width, global_bounds)
    chi2_per_dof = objective(sampled_params.tolist(), df_fixed, ratio_list)

    row = {
        "αs": float(np.round(sampled_params[0], 6)),
        "params": json.dumps(np.round(sampled_params[1:], 6).tolist()),
        "chi2/dof": float(chi2_per_dof),
        "muH_ratio": 1.0,
        "muJ_ratio": 1.0,
        "muRen_ratio": 1.0,
    }

    pd.DataFrame([row]).to_csv(out_path, mode="a", header=False, index=False)
```
