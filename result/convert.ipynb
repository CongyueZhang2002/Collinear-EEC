```python
import pandas as pd

df = pd.read_csv(
    '90_scan_1_old.csv',
    converters={'params': lambda s: [float(x) for x in s.strip('[]').split()]}
)

df['params'] = df['params'].apply(lambda p: [p[0], p[1]**(1/p[2]), p[2], p[3]])
df['params'] = df['params'].apply(lambda p: '[' + ' '.join(f'{x:.6f}' for x in p) + ']')

df.to_csv('90_scan_1.csv', index=False)
```
